```python
import pandas as pd
import numpy as np
from pymongo import MongoClient
```

```python
file_path = "nhs_conditions_v2.csv"
nhs_conditions_df = pd.read_csv(file_path)
```

```python
nhs_conditions_df.describe()
nhs_conditions_df.info()
```

```python
client = MongoClient("mongodb://localhost:27017/")
db = client["nhs_conditions_db"]
collection = db["conditions"]
```

```python
def insert_data_to_mongodb(df, collection):
    for index, row in df.iterrows():
        data = {
            "Condition_name": row["condition"],
            "Symptoms": row["symptoms"],
            "Causes": row["causes"],
            "Warnings": row["warnings"],
            "Recommendations": row["recommendations"]
        }
        collection.insert_one(data)
        print(f"Inserted data for condition: {row['condition']}")
    return "Data insertion completed successfully!"
```

```python
insert_data_to_mongodb(nhs_conditions_df, collection)
```
