```python
# 2. Start Ollama server in background
import subprocess
import time
import os

os.system("pkill -9 ollama")
time.sleep(2)

env = os.environ.copy()
env["OLLAMA_ORIGINS"] = "*"
env["OLLAMA_HOST"] = "0.0.0.0:11434"

ollama_process = subprocess.Popen(
    ["ollama", "serve"],
    stdout=subprocess.PIPE,
    stderr=subprocess.PIPE
)
time.sleep(5)
print("Ollama server started and ready to accept requests")
```

```python
from pyngrok import ngrok, conf
NGROK_TOKEN = "GET YOUR TOKEN FROM ngrok"
# ngrok.set_auth_token(NGROK_TOKEN)
conf.get_default().auth_token = NGROK_TOKEN

try:
    for t in ngrok.get_tunnels():
        ngrok.disconnect(t.public_url)
except Exception:
    pass

public_url = ngrok.connect(11434, host_header="localhost:11434").public_url

print("\n" + "="*50)
print(f"🌍 Public Ollama API URL: {public_url}")
print("="*50)
```
